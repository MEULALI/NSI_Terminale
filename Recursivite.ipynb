<span style="text-decoration:underline">Remarque</span> : il existe différentes manières d'écrire une fonction récursive.<br>
Pour la fonction précédente, on pourrait écrire :

```python
# Avec une variable résultat

def mystère(n):
    if n == 0:
        résultat = 0
    else:
        résultat = n + mystère(n-1)  
    return résultat
```

Dans ce programme, chaque appel a sa propre variable locale <code>résultat</code> qui peut contenir une valeur différente. 

```python
# Inversion des conditions

def somme(n):
    if n > 0:
        return n + somme(n-1)
    else:
        return 0
```

```python
# Plus court, on peut se passer du else

def mystère(n):
    if n == 0:
        return 0
    return n + mystère(n-1)
```

```python
# Encore plus court

def somme(n):
    return n if n==0 else n+somme(n-1)
```

```python
# Encore plus court, avec une fonction lambda (fonction anonyme)
# utile pour les compétitions de "code golf" dans lequelles
# l'objectif est d'écrire le programme le plus court possible en nombre de caractères

s=lambda n:0 if n==0 else n+s(n-1)
```

<div style="background-color:rgb(200,240,240);border:solid blue 1px; padding: 5px;">

**Conseils pour écrire une fonction récursive**
<ol>
    <li><strong>Commencer par le cas de base</strong><br>
    Identifier la situation la plus simple, que l’on peut résoudre immédiatement sans appel récursif (par exemple : chaîne vide, liste vide, n = 0, ...).<br>
        Écrire ce cas en premier dans la fonction.
        C’est la <strong>condition d’arrêt</strong> : sans elle, la récursion ne s’arrête jamais!</li>
    <li><strong>Décomposer le problème</strong><br>
        - Se demander : « Si je sais résoudre le problème sur un cas plus petit, comment m’en servir pour résoudre le cas général ? »<br>
        - Écrire le cas général en appelant la fonction sur une version réduite du problème (par exemple n−1, sous-liste, sous-chaîne, ...).
    </li>
    <li><strong>Avancer vers le cas de base</strong><br>
    Vérifier que chaque appel récursif rapproche du cas de base (sinon, risque de boucle infinie).</li>
    <li><strong>Ne pas réinventer la roue</strong><br>
        Utiliser le résultat de l’appel récursif comme si la fonction savait déjà faire le travail sur le cas plus petit.<br>
C’est le principe du « fais confiance à ta propre fonction ».
    </li>
    <li><strong>Vérifier avec de petits exemples</strong><br>
    Tester la fonction sur les plus petits cas possibles (0, 1, liste vide, etc.).<br>
Vérifier pas à pas comment la fonction se déroule.</li>
</ol>
</div>

<h2 style="text-decoration:underline;">4- Compléments sur la récursivité</h2>

**Le cas de base est indispensable**

Si on oublie de traiter le cas de base (la condition de terminaison), la fonction s'appelle indéfiniment et le programme ne s'arrête jamais. C'est l'équivalent d'une boucle while infinie.<br>
Voilà pourquoi, lorsqu'on écrit une fonction récursive,  il faut tout de suite identifier et traiter le cas de base pour être sûr de ne pas l'oublier!

```python
# ce programme provoque une erreur de récursion
def a():
    a()
a()
```

**Pile d'appels**

Dans l'exemple précédent, on n'a en réalité pas eu de boucle infinie. Au lieu de cela, le programme s'est arrêté avec le message d'erreur suivant : <em>RecursionError: maximum recursion depth exceeded</em>    

Par défaut, le nombre maximal d'appels récursifs en Python est 1000.<br>
On peut augmenter ce nombre maximal de la façon suivante :

```python
import sys

sys.setrecursionlimit(1500)
```

Pourquoi le nombre d'appels récursifs est-il limité?<br>
Lorsqu’un programme exécute des fonctions, il doit se souvenir :
- où reprendre l’exécution une fois la fonction terminée,
- quelles sont les variables locales propres à cet appel.

Pour cela, l’ordinateur utilise une structure de données particulière : la **pile d’appels** (ou call stack en anglais).<br>
Chaque fois qu’une fonction est appelée, l’ordinateur empile un nouveau bloc d’informations appelé cadre d’activation (stack frame).<br>
Ce cadre contient notamment :
- les paramètres reçus,
- les variables locales de la fonction,
- l’adresse de retour (c’est-à-dire l’endroit du programme où continuer après l’appel).

Quand la fonction se termine avec un return, son cadre est dépilé et on revient au cadre de la fonction appelante.
La machine réserve une zone spéciale de la mémoire pour stocker la pile d'appels. C'est pourquoi, par défaut, cette zone est limitée en taille.

Remarque : la structure de pile sera étudiée en détail dans un prochain chapitre.

On peut avoir un aperçu de la pile d'appels avec Tutor :

```python
from tutor import tutor

def mystère(n):
    if n == 0:
        résultat = 0
    else:
        résultat = n + mystère(n-1)  
    return résultat

mystère(5)

tutor()
```

<span style="text-decoration:underline">Remarque</span> : Ce style de programmation est appelé la **programmation fonctionnelle**.<br>

La récursivité joue un rôle central dans la programmation fonctionnelle.
Dans ce paradigme, on cherche souvent à éviter les boucles classiques (for, while) et les variables modifiables. La récursivité devient alors le moyen naturel d’exprimer la répétition ou l’itération.

C’est pour cette raison que les langages fonctionnels comme Lisp (un des plus anciens langages, créé dans les années 1960) ou OCaml (un langage fonctionnel moderne, utilisé en recherche et en enseignement) mettent la récursivité au cœur de leur syntaxe et de leur pratique.

Dans ces langages, on définit fréquemment des fonctions récursives pour parcourir des listes, traiter des arbres, ou construire des algorithmes de manière élégante.
Ils disposent aussi d’optimisations particulières, comme la récursion terminale, qui permet d’exécuter des fonctions récursives de manière aussi efficace que des boucles, sans risque de débordement de pile.
    
Inversement, on peut toujours se passer de la récursivité, mais parfois au prix d'un code plus compliqué à écrire.

**Avantages et inconvénients de la récursivité**

<span style="text-decoration:underline">Avantage</span> : Dans de nombreuses situations, le version récursive est plus intuitive et plus rapide à programmer (avec un peu d'habitude) que la version itérative. Le programme est généralement plus court à écrire.<br>
<span style="text-decoration:underline">Inconvénient</span> : Les programmes récursifs sont un peu moins rapides à l'exécution que leur équivalent itératif, car ils engendrent de nombreux appels de fonction qui surchargent un peu le processeur (pile d'appels, mémorisation des différents états et paramètres).

**Cas où la récursivité est inefficace**
