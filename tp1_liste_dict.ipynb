<div style="border:1pt solid slateblue; border-radius:5pt; color:slateblue; padding:3px; background-color: lightcyan"> 
    <h2> ✨ Type abstrait dictionnaire </h2>
    
Le **type abstrait dictionnaire** permet d'associer à une *clef* qui n'est pas forcément un entier, une *valeur*, on parle aussi de *tableau associatif*.
    
Voici une interface minimale :

|Opération|Signature|Description|
|---|---|---|
|creer_dico|creer_dico()|Renvoie un dictionnaire|
|valeur|valeur(dico, clef)|Accès direct en lecture à  la valeur associée à la clef fixée|
|ajouter|ajouter(dico, clef, valeur)|Accès direct en écriture à  la valeur associée à la clef fixée|
    
    
 <h2> ✨ Implémentation avec un tableau </h2>
    
On a présenté dans [l'exercice 2 du cours](https://frederic-junier.gitlab.io/tnsi-parc/T1_Structures_de_donnees/1_Structures_lineaires/C1_Types_abstraits/P1_Type_Abstrait/P1_Type_Abstrait/#interface-et-implementation) une implémentation du type abstrait Dictionnaire  où les couples (clef, valeur) sont stockés dans un tableau statique pouvant contenir un nombre maximal de clefs.
    
L'inconvénient est la complexité des opérations `valeur` et `ajouter` qui est proportionnelle au nombre de clefs stockées dans le pire des cas (clef en fin de tableau) ou dans le cas moyen.
    
    
     
 <h2> ✨ Implémentation avec une  table de hachage </h2>
    
Une **fonction de hachage** associe à une *clef* un nombre dans une plage de valeurs entières comprises entre $0$ et $m - 1$. L'image par une fonction de hachage est appelée *haché*.

L'entier $m$ est la taille d'un tableau statique.
    
![alt](https://capytale2.ac-paris.fr/web/sites/default/files/2023/07-14/9-45-53/fonction_hachage.png)
 
    
Dans une implémentation du  **type abstrait dictionnaire**  par une **table de hachage**, une **fonction de hachage** associe à une *clef* un index dans un tableau de taille fixée et la *valeur* associée à la *clef* est stockée dans la case correspondante du tableau. 
    
⚠️ On fait l'hypothèse  que le coût du calcul d'un haché est *constant*, en $O(1)$.
    
    
Dans l'exemple ci-dessous, la *clef* `'Eva'` a pour  haché $4$ et la *valeur* associée  est `'0745384813'`. On pourrait stocker juste la *valeur*, mais il est possible que deux *clefs* distinctes donnent le même *haché*, on parle alors de **collision**.
    
Pour résoudre les **collisions** une solution est de stocker dans la case du tableau une liste chaînée où sont rangées  les couples '(clef, valeur)' dont les clefs donnent le même  *haché*.
    
On choisit souvent un nombre premier pour la taille $m$ du tableau de stockage afin d'éviter les problèmes de diviseurs communs, qui augmentent  le risque de collisions.
    
Dans l'exemple ci-dessous, les *clefs* `'Eric'` et `'Bob'` ont pour  haché $2$ et sont stockées avec leur valeur associée dans une même liste chaînée. 
    
    
![alt](https://capytale2.ac-paris.fr/web/sites/default/files/2023/07-14/10-01-04/dico_chainage.svg)
